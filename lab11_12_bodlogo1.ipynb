Бодлого 1. Багш нарын мэдээлэл 1



```python
# Алхам 1: input.txt файлыг үүсгэх (Та энд өөрийн шалгахыг хүссэн оролтоо бичиж болно)
input_data = """Уламбаяр Төртогтох эрэгтэй МБУС МЗТ 99044079
Баттогтох Цэен-Ойдов эрэгтэй МБУС МЗТ 99003771
Сандагдорж Балдорж эрэгтэй МБУС МТ 88859959
Уртнасан Гантогтох эмэгтэй МБУС МЗТ 88081699
Алимаа Бор эмэгтэй НХУС ТТ 88169994
Тунгалаг Балжир эмэгтэй НХУС АХТ 80159991"""

with open("input.txt", "w", encoding="utf-8") as f:
    f.write(input_data)

# Алхам 2: Файлаас уншиж боловсруулах үндсэн код
# 1-р мөрөнд input.txt файлыг utf-8 өгөгдөл уншихаар нээж байна.
with open("input.txt", "r", encoding="utf-8") as f:
    # 2-р мөрөнд оролтын файлын мөр бүрийг уншиж mylines list рүү хадгалж байна.
    mylines = f.readlines()

# 3-р мөрөнд teachers = [] гэж багш нарын мэдээлэл хадгалах List үүсгэж байна.
teachers = []

# 4-р мөрөнд salbar = {} гэж салбар бүрийн багшийг тоолох Dict үүсгэж байна.
salbar = {}

# 6-13-р мөрөнд mylines list-ээр давталт гүйлж мөр бүрийг баганаар салгаж байна.
for line in mylines:
    parts = line.strip().split()
    if len(parts) >= 6:
        rd = {
            'ner': parts[0],
            'ovog': parts[1],
            'huis': parts[2],
            'salbar': parts[3],
            'tenhim': parts[4],
            'utas': parts[5]
        }
        teachers.append(rd)

        # 9-12-р мөрөнд хэрэв rd['salbar'] утгатай key үүссэн бол тоог 1-ээр нэмж, үүсээгүй бол 1 гэж оноож байна.
        s_name = rd['salbar']
        if s_name in salbar:
            salbar[s_name] += 1
        else:
            salbar[s_name] = 1

# 14-р мөрөнд гаралтын output.txt файлыг utf-8 өгөгдөл бичихээр нээж байна.
with open("output.txt", "w", encoding="utf-8") as f:
    # 15-16-р мөрөнд salbar dict объектийг key-р нь өсөхөөр эрэмбэлж хэвлэж байна.
    for key in sorted(salbar.keys()):
        line_to_write = f"{key} {salbar[key]}\n"
        f.write(line_to_write)

# Үр дүнг Google Colab дэлгэц дээр шууд харуулах
print("--- output.txt файлын агуулга ---")
with open("output.txt", "r", encoding="utf-8") as f:
    print(f.read())
```

Бодлого 2. Багш нарын мэдээлэл

```python
# 1. Жишээ оролтын өгөгдлийг input.txt файлд бичих (Та хүсвэл өөрийн файлаараа сольж болно)
sample_input = """Уламбаяр Төртогтох эрэгтэй МБУС МЗТ 99044079
Баттогтох Цэен-Ойдов эрэгтэй МБУС МЗТ 99003771
Сандагдорж Балдорж эрэгтэй МБУС МТ 88859959
Уртнасан Гантогтох эмэгтэй МБУС МЗТ 88081699
Алимаа Бор эмэгтэй НХУС ТТ 88169994
Тунгалаг Балжир эмэгтэй НХУС АХТ 80159991"""

with open('input.txt', 'w', encoding='utf-8') as f:
    f.write(sample_input)

# 2. Оператор тодорхойлох функц
def get_operator(phone):
    prefix = phone[:2]
    # Мобиком угтварууд
    if prefix in ['99', '94', '95', '85', '84']:
        return 'Мобиком'
    # Юнител угтварууд
    elif prefix in ['88', '89', '91', '96']:
        return 'Юнител'
    # Скайтел угтварууд
    elif prefix in ['90', '93', '98']:
        return 'Скайтел'
    # Жимобайл угтварууд (80, 81 болон 86, 87 гэх мэт)
    elif prefix in ['86', '87', '80', '81']:
        return 'Жимобайл'
    else:
        return 'Бусад'

# 3. input.txt файлыг унших
with open('input.txt', 'r', encoding='utf-8') as f:
    mylines = f.readlines()

operator_counts = {}

# Мөр бүрээр ялгаж утасны дугаарыг олох
for line in mylines:
    parts = line.strip().split()
    if len(parts) >= 6:
        phone = parts[5] # Сүүлийн багана нь утасны дугаар
        op = get_operator(phone)

        if op in operator_counts:
            operator_counts[op] += 1
        else:
            operator_counts[op] = 1

# Хамгийн олон хэрэглэгчтэй операторыг олох (max утгаар нь шүүх)
if operator_counts:
    max_operator = max(operator_counts, key=operator_counts.get)
    max_count = operator_counts[max_operator]
    result_str = f"{max_operator} {max_count}"
else:
    result_str = "Өгөгдөл олдсонгүй"

# 4. output.txt файл руу бичих болон хэвлэх
with open('output.txt', 'w', encoding='utf-8') as f_out:
    f_out.write(result_str + "\n")
    print("Гаралт:")
    print(result_str)

# output.txt файлын агуулгыг шалгах
print("\n--- output.txt файлын агуулга ---")
with open('output.txt', 'r', encoding='utf-8') as f_out:
    print(f_out.read())
```

Бодлого 3: Бутархайг хураах

```python
# 1. Хоёр тооны ХИЕХ (Хамгийн Их Ерөнхий Хуваагч) олох рекурсив эсвэл энгийн функц
def hieh(a, b):
    while b != 0:
        a, b = b, a % b
    return a

# 2. Бутархайг хураах функц
def reduce_fraction(a, b):
    # Хүртвэр болон хуваарийн ХИЕХ-ийг олно
    common = hieh(abs(a), abs(b))

    # Хүртвэр болон хуваарийг ХИЕХ-д хувааж хураана
    reduced_a = a // common
    reduced_b = b // common

    return reduced_a, reduced_b

# --- Жишээ тестээр шалгах ---
# Жишээ нь: 24/36 бутархайг хураавал 2/3 гарна
a = 24
b = 36

res_a, res_b = reduce_fraction(a, b)

print(f"Өгөгдсөн бутархай: {a} / {b}")
print(f"Хураасан бутархай: {res_a} / {res_b}")

# Хэрэв файжтай ажиллах хэлбэрээр (input.txt / output.txt) өгөгдсөн бол:
"""
# input.txt файлд a болон b зайгаар тусгаарлагдан өгөгдсөн гэж үзвэл:
with open('input.txt', 'r', encoding='utf-8') as f:
    line = f.readline().strip()
    a, b = map(int, line.split())

res_a, res_b = reduce_fraction(a, b)

with open('output.txt', 'w', encoding='utf-8') as f_out:
    f_out.write(f"{res_a} {res_b}\\n")
"""
```

Бодлого 4: Фибоначийн тоон дараалал

```python
def check_fibonacci(n):
    if n < 0:
        return False, 0

    # Фибоначийн дарааллыг үүсгэж шалгах
    a, b = 0, 1
    index = 1

    if n == 0:
        return True, 0 # 0-р гишүүн эсвэл эхний гэж үзэж болно

    while b <= n:
        if b == n:
            return True, index
        a, b = b, a + b
        index += 1

    return False, 0

# Шалгах жишээ
num = 8  # Жишээ нь 8 бол 6 дахь гишүүн (1, 1, 2, 3, 5, 8)
is_fib, idx = check_fibonacci(num)

if is_fib:
    print(f"Бодлого 6_04: {num} нь Фибоначийн тоо мөн. Энэ нь {idx}-р гишүүн байна.")
else:
    print(f"Бодлого 6_04: {num} нь Фибоначийн тоо биш байна.")
```

Бодлого 5: Тэгш өнцөгтийн класс (Rectangle)

```python
class Rectangle:
    def __init__(self, *args):
        # 1. Өөр объектоос хуулж авах (Хэрэв эхний аргумент нь Rectangle объект байвал)
        if len(args) == 1 and isinstance(args[0], Rectangle):
            other = args[0]
            self.a = other.a
            self.b = other.b
            self.name = other.name + "_copy"

        # 2. Параметраас авдаг (Rectangle(a, b, name))
        elif len(args) == 3:
            self.a = args[0]
            self.b = args[1]
            self.name = args[2]

        # 3. Файлаас авдаг (Rectangle('filename.txt'))
        elif len(args) == 1 and isinstance(args[0], str):
            filename = args[0]
            with open(filename, 'r', encoding='utf-8') as f:
                line = f.readline().strip().split()
                self.a = float(line[0])
                self.b = float(line[1])
                self.name = line[2]

        # 4. Гараас уншдаг (Аргументгүй үед)
        else:
            self.a = float(input("Тэгш өнцөгтийн урт (a) оруулна уу: "))
            self.b = float(input("Тэгш өнцөгтийн өргөн (b) оруулна уу: "))
            self.name = input("Тэгш өнцөгтийн нэр оруулна уу: ")

    def area(self):
        return self.a * self.b

    def perimeter(self):
        return 2 * (self.a + self.b)

    def display(self):
        print(f"Нэр: {self.name}, Урт: {self.a}, Өргөн: {self.b}, Талбай: {self.area()}, Периметр: {self.perimeter()}")

# --- Туршиж үзэх (4 хувилбараар объект үүсгэх) ---
# Туршилтын зорилгоор файлд өгөгдөл бэлдэх
with open('rect_input.txt', 'w', encoding='utf-8') as f:
    f.write("5.5 4.0 FileRect")

print("--- Бодлого 6_05 үр дүнгүүд ---")
# 1. Параметраас авсан объект
r1 = Rectangle(10, 5, "ParamRect")
r1.display()

# 2. Файлаас авсан объект
r2 = Rectangle('rect_input.txt')
r2.display()

# 3. Өөр объектоос хуулж авсан объект
r3 = Rectangle(r1)
r3.display()

# 4. Гараас унших объект (Кодад зориулж түр коммент болгож болно, эсвэл ажиллуулаад гараас утга өгнө)
# r4 = Rectangle()
# r4.display()
```

Бодлого 6: Квадрат тэгшитгэлийн класс (EqQuad)

```python
import math

class EqQuad:
    def __init__(self, *args):
        # 1. Өөр объектоос хуулж авах
        if len(args) == 1 and isinstance(args[0], EqQuad):
            other = args[0]
            self.a = other.a
            self.b = other.b
            self.c = other.c

        # 2. Параметраас авдаг (EqQuad(a, b, c))
        elif len(args) == 3:
            self.a = args[0]
            self.b = args[1]
            self.c = args[2]

        # 3. Файлаас авдаг (EqQuad('eq_input.txt'))
        elif len(args) == 1 and isinstance(args[0], str):
            filename = args[0]
            with open(filename, 'r', encoding='utf-8') as f:
                line = f.readline().strip().split()
                self.a = float(line[0])
                self.b = float(line[1])
                self.c = float(line[2])

        # 4. Гараас уншдаг
        else:
            self.a = float(input("a коэффицентийг оруулна уу: "))
            self.b = float(input("b коэффицентийг оруулна уу: "))
            self.c = float(input("c коэффицентийг оруулна уу: "))

    def bodolt(self):
        print(f"Тэгшитгэл: {self.a}x^2 + {self.b}x + {self.c} = 0")
        if self.a == 0:
            if self.b == 0:
                print("Шийдгүй эсвэл төгсгөлгүй олон шийдтэй.")
            else:
                x = -self.c / self.b
                print(f"Шугаман тэгшитгэлийн шийд: x = {x}")
            return

        d = self.b**2 - 4 * self.a * self.c
        if d > 0:
            x1 = (-self.b + math.sqrt(d)) / (2 * self.a)
            x2 = (-self.b - math.sqrt(d)) / (2 * self.a)
            print(f"Хоёр язгууртай: x1 = {x1}, x2 = {x2}")
        elif d == 0:
            x = -self.b / (2 * self.a)
            print(f"Нэг давхар язгууртай: x = {x}")
        else:
            real_part = -self.b / (2 * self.a)
            imag_part = math.sqrt(abs(d)) / (2 * self.a)
            print(f"Комплекс язгууртай: x1 = {real_part} + {imag_part}i, x2 = {real_part} - {imag_part}i")

# --- Туршиж үзэх (4 хувилбараар объект үүсгэх) ---
# Файлын жишээ бэлдэх
with open('eq_input.txt', 'w', encoding='utf-8') as f:
    f.write("1 -3 2") # x^2 - 3x + 2 = 0 (Язгуурууд нь 1 ба 2)

print("\n--- Бодлого 6_06 үр дүнгүүд ---")
# 1. Параметраас авсан объект
eq1 = EqQuad(1, -5, 6)
eq1.bodolt()

# 2. Файлаас авсан объект
eq2 = EqQuad('eq_input.txt')
eq2.bodolt()

# 3. Өөр объектоос хуулж авсан объект
eq3 = EqQuad(eq1)
eq3.bodolt()

# 4. Гараас унших объект (Хүсвэл доодохыг идэвхжүүлж болно)
# eq4 = EqQuad()
# eq4.bodolt()
```
