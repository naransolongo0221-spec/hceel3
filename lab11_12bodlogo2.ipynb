```python
# Алхам 1: input.txt файлыг үүсгэх (Энд жишээ өгөгдлийг оруулсан болно)
input_data = """Гантогтох Дашхүү Эрэгтэй АХБС ОХТ 99806180
Нямгэрэл Сосор Эмэгтэй НХУС БиоТ 91791910
Буянтогтох Сосор Эрэгтэй БоСС Т7 90872378
Гармаа Цэен-Ойдов Эмэгтэй БоСС Т1 98492228
Төрбат Дорж Эрэгтэй БС Т9 86628526
Баттогтох Цэен-Ойдов Эрэгтэй БТС ГаТ 93445533
Буянтогтох Тогтох Эмэгтэй СӨБС Т7 91594219
Баттогтох Ганбаяр Эмэгтэй Лицей МАТ 80669800
Нямсүрэн Ренчин Эрэгтэй БС Т1 91694985
Гоомарал Бор Эмэгтэй БоСС АХС 80271984"""

with open("input.txt", "w", encoding="utf-8") as f:
    f.write(input_data)

# Утасны дугаарын эхний 2 оронгоор оператор тодорхойлох функц
def get_operator(phone):
    phone_prefix = phone[:2]
    # Мобиком: 99, 94, 95, 85
    if phone_prefix in ['99', '94', '95', '85']:
        return 'Мобиком'
    # Юнител: 88, 89, 91, 96
    elif phone_prefix in ['88', '89', '91', '96']:
        return 'Юнител'
    # Скайтел: 90, 93, 98
    elif phone_prefix in ['90', '93', '98']:
        return 'Скайтел'
    # Жимобайл: 80, 86, 87 (зарим 80-ууд Жимобайл эсвэл өөр байж болох тул ерөнхий стандартаар тохируулна)
    elif phone_prefix in ['80', '86', '87']:
        return 'Жимобайл'
    else:
        return 'Бусад'

# Алхам 2: input.txt файлыг унших
with open("input.txt", "r", encoding="utf-8") as f:
    mylines = f.readlines()

operator_counts = {}

# Мөр бүрээс мэдээллийг задалж утасны дугаараар операторыг тоолох
for line in mylines:
    parts = line.strip().split()
    if len(parts) >= 6:
        utas = parts[5]
        op = get_operator(utas)

        if op in operator_counts:
            operator_counts[op] += 1
        else:
            operator_counts[op] = 1

# Хамгийн олон хэрэглэгчтэй үүрэн холбооны компани болон тоог олох
if operator_counts:
    max_op = max(operator_counts, key=operator_counts.get)
    max_count = operator_counts[max_op]
else:
    max_op = "Тодорхойгүй"
    max_count = 0

# Алхам 3: output.txt файлд бичих
with open("output.txt", "w", encoding="utf-8") as f:
    f.write(f"{max_op} {max_count}\n")

# Үр дүнг Google Colab дэлгэц дээр харуулах
print("--- output.txt файлын агуулга ---")
with open("output.txt", "r", encoding="utf-8") as f:
    print(f.read())
```

Бодлого 6: Фибоначийн тоон дараалал

```python
# Алхам 1: input.txt файлыг үүсгэх (Жишээ тоонууд: 8, 13, 7 гэх мэт)
input_data = """8
13
7
21"""

with open("input.txt", "w", encoding="utf-8") as f:
    f.write(input_data)

# Лекц 6-ийн санааг ашиглан Фибоначийн гишүүдийг үүсгэж, шалгах функц
def check_fibonacci(n):
    if n < 0:
        return f"{n} - Фибоначийн тоо биш"

    # 0 болон 1-ийн хувьд тусгайлан шалгах
    if n == 0:
        return f"0 нь 0 дэх гишүүн"
    if n == 1:
        return f"1 нь 1 дэх гишүүн (эсвэл 2 дахь гишүүн)" # 1 хоёр удаа тохиолддог тул эхнийхийг нь авч болно

    a, b = 0, 1
    index = 1

    while b < n:
        a, b = b, a + b
        index += 1

    if b == n:
        return f"{n} нь {index} дэх гишүүн"
    else:
        return f"{n} - Фибоначийн тоо биш"

# Алхам 2: input.txt файлыг уншиж шалгах
with open("input.txt", "r", encoding="utf-8") as f:
    lines = f.readlines()

output_lines = []

for line in lines:
    line_str = line.strip()
    if line_str:
        num = int(line_str)
        result = check_fibonacci(num)
        output_lines.append(result + "\n")

# Алхам 3: output.txt файлд бичих
with open("output.txt", "w", encoding="utf-8") as f:
    f.writelines(output_lines)

# Үр дүнг Google Colab дэлгэц дээр харуулах
print("--- output.txt файлын агуулга ---")
with open("output.txt", "r", encoding="utf-8") as f:
    print(f.read())
```
